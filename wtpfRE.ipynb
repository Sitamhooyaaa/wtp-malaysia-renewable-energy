```python
# Importing required library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
%matplotlib inline

print("Libraries loaded successfully")
```

```python
df = pd.read_excel('wtp-data.xlsx')
print(f"Dataset shape: {df.shape}")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

```python
# How many missing values per column
missing = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': missing,
    'Percentage': missing_percent
})

# Only show columns that have missing values
missing_summary = missing_summary[missing_summary['Missing Values'] > 0]
print(missing_summary.sort_values('Percentage', ascending=False))
```

Drop the columns with more than 40% because there are too much missing values. Because if we tried to fill it, it will cause our data to be learning from fake data.
And somemore, the missing column that have for example more than 89% missing values carries no statistical information. 

```python
# Drop columns with too many missing values (>40% threshold)
cols_to_drop = [
    'others - education',  # 93% missing
    'maxwtp2_y',           # 75% missing
    'maxwtp2_n',           # 76% missing
    'maxwtp1_y',           # 74% missing
    'maxwtp1_n',           # 71% missing
    'wtp2_ori',            # 48% missing
    'bid2_or',             # 45% missing
]

df = df.drop(columns=cols_to_drop)
print(f"Columns remaining: {df.shape[1]}")
```

```python
# Re-check remaining missing values
missing = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': missing,
    'Percentage': missing_percent
})

# Only show columns that have missing values
missing_summary = missing_summary[missing_summary['Missing Values'] > 0]
print(missing_summary.sort_values('Percentage', ascending=False))
```

```python
# Fill missing values for income columns with 0
# why with 0 not mean/median? Because missing income likely means no income, especially for children and possibly for unemployed spouses. Filling with 0 is more realistic than using mean/median which could inflate the income values.
df['husband income'] = df['husband income'].fillna(0)
df['wife income'] = df['wife income'].fillna(0)
df['children income'] = df['children income'].fillna(0)

# Fill d101, d102, d104 with median
# why median? Because these are likely ordinal or continuous variables that could be skewed, and median is more robust to outliers than mean.
df['d101'] = df['d101'].fillna(df['d101'].median())
df['d102'] = df['d102'].fillna(df['d102'].median())
df['d104'] = df['d104'].fillna(df['d104'].median())

# Drop wtp_ori - only 1 missing and it's a duplicate of wtp anyway
df = df.drop(columns=['wtp_ori'])

# Verify no more missing values
print(f"Remaining missing values: {df.isnull().sum().sum()}")
print(f"Dataset shape: {df.shape}")
```

Next, removing the column that is not useful for our modelling. Also, there are columns that have model output since this data is downloaded from kaggle, so we remove that too. 

```python
# Drop columns that shouldn't be used in modelling
cols_to_remove = [
    'idall',        # Just an ID number, no information
    'state ID',     # Duplicate of state column
    'wtp1',         # Direct follow up to wtp, would leak target
    'wtp2',         # Same reason
    'bid2',         # Follow up bid, depends on wtp response
    'maxwtp',       # Maximum wtp, directly derived from target
    'yy', 'yn', 'ny', 'nn',  # These are wtp response breakdowns
    'neg_dumm1', 'neg_dumm2', 'neg_dumm3',  # Derived from wtp
    'pmodel1', 'pmodel2', 'pmodel6',         # Model outputs already in data
    '_est_model1', '_est_model2', 
    '_est_model3', '_est_fmodel',            # Same, model estimates
    'confident1',   # Derived variable
    'bilres',       # Internal survey reference
]

df = df.drop(columns=cols_to_remove)
print(f"Final shape: {df.shape}")
print(f"\nRemaining columns:")
print(df.columns.tolist())
```

After that, we will decide the target variable and seperate it with the other features. 

```python
# Define target variable
target = 'wtp'

# Define features (everything except target)
X = df.drop(columns=[target])
y = df[target]

print(f"Features shape: {X.shape}")
print(f"Target shape: {y.shape}")
print(f"\nTarget value counts:")
print(y.value_counts())
print(f"\nTarget percentage:")
print(y.value_counts(normalize=True) * 100)
```
